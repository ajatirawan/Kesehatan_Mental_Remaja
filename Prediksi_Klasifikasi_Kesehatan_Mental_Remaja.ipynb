```python
#Melakukan split data latih dan data uji
from sklearn.model_selection import train_test_split

x_latih, x_uji, y_latih, y_uji = train_test_split(fitur, target, test_size=0.3)
#x untuk fitur dan y untuk latih
#x_latih ini 80 persen dari data training dan y latih 80 persen
#y_uji ini 20 % dari data uji dan dari y data uji (y_uji)
#test nya sebesar 20 % = 0.2 dari 100 %
```

```python
# Klasifikasi Menggunakan Algoritma LogisticRegression (Untuk Klasifikasi)
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression
clf = LogisticRegression(max_iter = 1000) #Data 1200
clf = clf.fit(x_latih, y_latih) #latih dengan data latih + label latih, fitting
y_pred = clf.predict(x_uji)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

akurasi = accuracy_score(y_uji, y_pred)
print("Akurasi:", akurasi)

print("\nConfusion Matrix:")
print(confusion_matrix(y_uji, y_pred))

# Laporan klasifikasi (precision, recall, f1-score)
print("\nClassification Report:")
print(classification_report(y_uji, y_pred))
```

```python
from imblearn.over_sampling import SMOTE
from collections import Counter
```

```python
# ==========================================
# SMOTE UNTUK MENANGANI IMBALANCE DATA
# ==========================================

from imblearn.over_sampling import SMOTE

print("Distribusi kelas sebelum SMOTE:")
print(y_latih.value_counts())

# Terapkan SMOTE pada data training
smote = SMOTE(random_state=42)

x_latih_smote, y_latih_smote = smote.fit_resample(
    x_latih,
    y_latih
)

print("\nDistribusi kelas setelah SMOTE:")
print(y_latih_smote.value_counts())
```

```python
fig, ax = plt.subplots(1,2, figsize=(12,5))

sns.countplot(
    x=y_latih,
    ax=ax[0],
    palette='Set2'
)
ax[0].set_title("Sebelum SMOTE")

sns.countplot(
    x=y_latih_smote,
    ax=ax[1],
    palette='Set2'
)
ax[1].set_title("Sesudah SMOTE")

plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# Model Logistic Regression
model_lr_smote = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Training dengan data hasil SMOTE
model_lr_smote.fit(
    x_latih_smote,
    y_latih_smote
)

# Prediksi data uji
y_pred_lr_smote = model_lr_smote.predict(x_uji)

# Evaluasi
print(classification_report(y_uji, y_pred_lr_smote))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Membuat Confusion Matrix
cm = confusion_matrix(y_uji, y_pred_lr_smote)

print("Confusion Matrix:")
print(cm)

# Visualisasi Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Tidak Depresi', 'Depresi'],
    yticklabels=['Tidak Depresi', 'Depresi']
)

plt.title('Confusion Matrix Logistic Regression + SMOTE')
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

# Model Decision Tree
model_dt_smote = DecisionTreeClassifier(random_state=42)

# Training dengan data hasil SMOTE
model_dt_smote.fit(
    x_latih_smote,
    y_latih_smote
)

# Prediksi
y_pred_dt_smote = model_dt_smote.predict(x_uji)
```

```python
from sklearn.metrics import classification_report

print("=== Decision Tree + SMOTE ===")
print(classification_report(y_uji, y_pred_dt_smote))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm_dt = confusion_matrix(y_uji, y_pred_dt_smote)

print("Confusion Matrix Decision Tree + SMOTE:")
print(cm_dt)

plt.figure(figsize=(6,5))
sns.heatmap(
    cm_dt,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Tidak Depresi', 'Depresi'],
    yticklabels=['Tidak Depresi', 'Depresi']
)

plt.title('Confusion Matrix Decision Tree + SMOTE')
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.show()
```
